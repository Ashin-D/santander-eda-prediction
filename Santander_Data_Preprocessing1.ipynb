```python

import pandas as pd
import numpy as np

df = pd.read_csv('/content/train.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
df.head()
```

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes.value_counts())

print("\nFirst five rows:")
display(df.head())

print("\nDataset information:")
df.info()
```

```python

missing = df.isnull().sum()

print("Total missing values:", missing.sum())

print("\nColumns containing missing values:")
print(missing[missing > 0])
```

```python

numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

print("Missing values after cleaning:",
      df.isnull().sum().sum())
```

```python

print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Shape after removing duplicate rows:", df.shape)
```

```python

constant_cols = [
    col for col in df.columns
    if col != 'TARGET' and df[col].nunique(dropna=False) <= 1
]

print("Number of constant columns:", len(constant_cols))
print("Constant column names:", constant_cols)

df = df.drop(columns=constant_cols)

print("Shape after removing constant columns:", df.shape)
```

```python

feature_cols = [
    col for col in df.columns if col != 'TARGET'
]

duplicate_cols = df[feature_cols].T.duplicated()
duplicate_cols = duplicate_cols[duplicate_cols].index.tolist()

print("Number of duplicate feature columns:",
      len(duplicate_cols))
print("Duplicate columns:", duplicate_cols)

df = df.drop(columns=duplicate_cols)

print("Shape after removing duplicate columns:", df.shape)
```

```python

print("Summary of numerical features:")
display(df.describe().T)

print("\nNumber of unique values per column:")
print(df.nunique().sort_values().head(20))

if 'var3' in df.columns:
    print("\nMost frequent values in var3:")
    print(df['var3'].value_counts().head(10))
```

```python
print("Columns with only 1 unique value:")
print(df.columns[df.nunique() <= 1].tolist())
```

```python
df['var3'] = df['var3'].replace(-999999, np.nan)

median_var3 = df['var3'].median()

df['var3'] = df['var3'].fillna(median_var3)

print("Remaining missing values:", df.isnull().sum().sum())
print("var3 median used:", median_var3)
```

```python
print("Final dataset shape:", df.shape)

print("Remaining constant columns:",
      df.columns[df.nunique() <= 1].tolist())

print("Duplicate rows:", df.duplicated().sum())
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/train.csv')

print("Dataset restored:", df.shape)
```

```python
# Handle the special value in var3
df['var3'] = df['var3'].replace(-999999, np.nan)
df['var3'] = df['var3'].fillna(df['var3'].median())

# Fill any remaining missing numeric values
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

print("Missing values remaining:", df.isnull().sum().sum())
print("Dataset shape:", df.shape)
```

```python
constant_cols = [
    col for col in df.columns
    if df[col].nunique() <= 1
]

print("Constant columns found:", len(constant_cols))

df = df.drop(columns=constant_cols)

print("Shape after removing constant columns:", df.shape)
print("Rows preserved:", len(df))
```

```python
feature_cols = [col for col in df.columns if col != 'TARGET']

duplicate_cols = df[feature_cols].T.duplicated()
duplicate_cols = duplicate_cols[duplicate_cols].index.tolist()

print("Duplicate feature columns found:", len(duplicate_cols))

df = df.drop(columns=duplicate_cols)

print("Shape after removing duplicate columns:", df.shape)
print("Rows preserved:", len(df))
```
